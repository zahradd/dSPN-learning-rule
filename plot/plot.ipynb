```python
path=r"C:\Users\zahra.khodadadi\Documents\GitHub\plasticity\experiments\cluster_0\data"
fileName='test'
import os
import matplotlib.pyplot as plt
import numpy as np
from scipy.signal import argrelextrema, find_peaks
import numpy as np
import json
import sys
import seaborn as sns
import neuron as nrn
from neuron  import h
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import FuncFormatter
from matplotlib import rcParams
from operator import sub
import os
import json
import os

import pickle
from neuron  import h
import numpy as np
import neuron as nrn
from scipy.signal import argrelextrema, find_peaks
import seaborn as sns
import os
import pickle
import matplotlib.pyplot as plt
import numpy as np
import sys
def load_obj(name ):
    '''
    functions used to load data in the pickle format. 
    Used to un-pickle data from dictionaries
    
    name    = file name of the pickled data, including ending
    '''
    
    with open(name, 'rb') as f:
        return pickle.load(f) 


def load_all_objects(path, fileName):
    loaded_data = {}  # Start with a blank dictionary
    
    suffixes = [
        '_WFirst', '_Wend', '_WInFirst', '_WInEnd', '_performance', '_secNames','_Rec_Volt_Dend',
        '_tms', '_tresh', '_calMaxs', '_train', '_treshs_in', '_treshsmin_in', '_calMaxs_in',
        '_ltype','_Rec_time', '_Rec_Volt'
    ]
    
    for suffix in suffixes:
        try:
            loaded_data[suffix[1:]] = load_obj(os.path.join(path, f"{fileName}{suffix}.pkl"))
        except FileNotFoundError:
            print(f"File for {suffix[1:]} not found.")
            loaded_data[suffix[1:]] = None  # or any default value you'd like to assign
    
    return loaded_data

# Usage

loaded_data = load_all_objects(path, fileName)

vms=[]
# Access individual variables
train = loaded_data['train']  # Access 'train' data
performance = loaded_data['performance']  # Access 'performance' data
tms = loaded_data['tms']  # Access 'tms' data
secNames = loaded_data['secNames']  # Access 'secNames' data
tresh = loaded_data['tresh']  # Access 'tresh' data
calMaxs = loaded_data['calMaxs']  # Access 'calMaxs' data
Wend = loaded_data['Wend']  # Access 'Wend' data
WFirst = loaded_data['WFirst']  # Access 'WFirst' data
WInFirst = loaded_data['WInFirst']  # Access 'WInFirst' data
WInEnd = loaded_data['WInEnd']  # Access 'WInEnd' data
treshs_in = loaded_data['treshs_in']  # Access 'treshs_in' data
treshsmin_in = loaded_data['treshsmin_in']  # Access 'treshsmin_in' data
calMaxs_in = loaded_data['calMaxs_in']  # Access 'calMaxs_in' data
ltype = loaded_data['ltype']  # Access 'ltype' data
Rec_time = loaded_data['Rec_time']  # Access 'tmRs' data
Rec_Volt = loaded_data['Rec_Volt']  # Access 'vms' data
Rec_Volt_Dend = loaded_data['Rec_Volt_Dend']  # Access 'Rec_Vol_Dends' data

```
